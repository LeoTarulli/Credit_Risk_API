# Credit Risk Project 

Understanding our Dataset

* checking_status: The currently balance account.
* duration: duration of the credit payment, in months.
* credit_history: Is the customer credit history. It can have the following values:
* existing paid: currently credit process
* all paid: all the credit abre paid by the customer
* delayed previously: past delays at # the credit payment
* critical/ other existing credit: critical credit accounts
* no credits/ all paid: customer that never gets credit.
* purpose: porpouse of the credit requested.
* credit_amount: Amount of the credit requested
* savings_status: Balance account of savings/ bond.
* employment:How many years the customer has employed.
* installment_commitment: Percentual that the customer cannot reach due to the income.

--> When a customer takes a credit, the installment cannot ultrapass a percentual of customer income.
* personal_status: if is divorced, married, etc.
* other_parties: Another part involde on th credit, like guarantos ("fiador").
* residence_since: How many year the customer has in the country.
* property_magnitude: The good of the customer.
* age: The age of the customer.
* other_payment_plans: If the customer has another credit line and how the type (store, bank, etc).
* housing:The housing customer type, could be own, for free or rent.
* existing_credits: Number of existing credits at this bank.
* job: This represents whats type of job the customer has. Could be skilled, unskilled resident, high qualif/self emp/mgmt and unemp/unskilled non res.
* num_dependents: The number of depedents that live wth the customer.
* own_telephone: If the customer has a telephone.
* foreign_worker: If the customer is a foreign or not.
* class: Our target. If the credit is goos or bad to the bank.

## Face 3: Machine Learning with Hyperparameter tuning

At this stage of the project, I will build several classification machine learning models to predict if a applicat is good or bad but firstly we will transform our categorical data into numerical and we will group them into:
* ordinal (ranked),
* nominal (unranked),
* and target columns

```python
from sklearn.metrics import accuracy_score

# 1. Obtaining our string columns
categorical_columns = df.select_dtypes(include="string").columns

for i in categorical_columns:
    print(pd.DataFrame(df[i].value_counts()))
```

```python
# 1. Target Column (bad = 1, good = 0 for risk detection)
df['class'] = df['class'].map({'good': 0, 'bad': 1})

# 2. Ordinal / Ranked Features (True hierarchy)
ordinal_mappings = {
    'checking_status': {'no checking': 0, '<0': 1, '0<=X<200': 2, '>=200': 3},
    'savings_status': {'no known savings': 0, '<100': 1, '100<=X<500': 2, '500<=X<1000': 3, '>=1000': 4},
    'employment': {'unemployed': 0, '<1': 1, '1<=X<4': 2, '4<=X<7': 3, '>=7': 4},
    'job': {'unemp/unsk/ nonres': 0, 'unskilled': 1, 'skilled': 2, 'highqualif / selfemp': 3},
    'property_magnitude': {'no known property': 0, 'car': 1, 'life insurance': 2, 'real estate': 3}
}

for col, mapping in ordinal_mappings.items():
    df[col] = df[col].map(mapping)

# 3. Binary Features
df['foreign_worker'] = df['foreign_worker'].map({'no': 0, 'yes': 1})
df['own_telephone'] = df['own_telephone'].map({'none': 0, 'yes': 1})

# 4. Nominal / Unranked Features (One-Hot Encoded)
nominal_cols = [
    'credit_history',
    'purpose',
    'personal_status',
    'other_parties',
    'other_payment_plans',
    'housing'
]

df = pd.get_dummies(df, columns=nominal_cols, drop_first=True, dtype=int)

# 5. Transofmring float types columns into integers
df = df.astype("int")
```

```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import warnings
warnings.filterwarnings("ignore")

import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme()

df = pd.read_csv("Clean_data",index_col = "Unnamed: 0")
```

```python
# Importing libraries
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split

# Setting our target and independents columns
y= df["class"]
X = df.drop(columns= "class")


X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.3, random_state = 42,  stratify=y)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

## Decision Tree Model - Credit Risk Project

```python
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay

dt_model = DecisionTreeClassifier(random_state = 42)

# the tunning parameters
param_grid_dt = {
    "criterion":["gini", "entropy"],
    "max_depth": [None,2,5,7,10,20],
    "min_samples_split":[2,3,4],
    "class_weight": [
        "balanced",
        {0: 1, 1: 1.5},  # Ponderación suave
        {0: 1, 1: 2.0},  # Error en 'Bad' cuesta el doble
        {0: 1, 1: 2.5},  # Error en 'Bad' cuesta 2.5 veces más
        {0: 1, 1: 3.0},  # Error en 'Bad' cuesta el triple
    ]
}

# Building the grid model
grid_dt = GridSearchCV(
    estimator = dt_model,
    param_grid = param_grid_dt,
    cv = 5,
    scoring="accuracy",
    n_jobs=-1,
    verbose=1)

# Training the model
grid_dt.fit(X_train,y_train)

best_dt = grid_dt.best_estimator_ # Getting the best estimators
DTC_pred = best_dt.predict(X_test) # Predicting y_test by using the best model

# Results
print("Best parameters: ", grid_dt.best_params_)
print("Best CV score: ", round(grid_dt.best_score_,2))
print(f"Accuracy :{round(accuracy_score(y_test, DTC_pred),2)}")


# Classification Report and Confusion Matrix
print("\nClassification Report:\n", classification_report(y_test, DTC_pred,  target_names = ["Good (0)", "Bad (1)"]))

cm = confusion_matrix(y_test, DTC_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix = cm,
    display_labels = ["Good (0)", "Bad (1)"]
)

disp.plot(cmap = "Blues", values_format = "d")
plt.grid(False)
plt.show()
```

## LogisticRegression - Hyperparameter tuning

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

lgr_model = LogisticRegression(random_state=42)

# the tunning parameters
param_grid_dt = {
    "max_iter":[100,150,200],
    "n_jobs": [5,7,10,20],
    "solver":['lbfgs','liblinear'],
    "class_weight": [
        "balanced",
        {0: 1, 1: 1.5},  
        {0: 1, 1: 2.0},  
        {0: 1, 1: 2.5},  
        {0: 1, 1: 3.0},  ]}

# Building the grid model
grid_lgr = GridSearchCV(
    estimator = lgr_model,
    param_grid = param_grid_dt,
    cv = 5,
    scoring="accuracy")

# Training the model
grid_lgr.fit(X_train,y_train)

best_lgr = grid_lgr.best_estimator_ # Getting the best estimators
LGR_pred = best_lgr.predict(X_test) # Predicting y_test by using the best model


# Results
print("Best CV score: ", round(grid_lgr.best_score_,2))
print(f"Accuracy :{round(accuracy_score(y_test, LGR_pred),2)}")


# Classification Report and Confusion Matrix
print("\nClassification Report:\n", classification_report(y_test, LGR_pred,  target_names = ["Good (0)", "Bad (1)"]))

cm = confusion_matrix(y_test, LGR_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix = cm,
    display_labels = ["Good (0)", "Bad (1)"]
)

disp.plot(cmap = "Blues", values_format = "d")
plt.grid(False)
plt.show()
```

## GaussianNB Model - Hyperparameter tuning

```python
# Importing the libraries
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, roc_auc_score, classification_report, ConfusionMatrixDisplay, confusion_matrix

# Importing the model
gnb = GaussianNB()

# Setting paramters
gnb_param = {"var_smoothing": np.logspace(0, -9, num=100)}

# Building the grid search
grid_gaussianNB = GridSearchCV(estimator=gnb, param_grid=gnb_param, cv=5, scoring='recall')

# Training the model
grid_gaussianNB.fit(X_train,y_train)

# Obtainig the best estimator and predicing y_test
best_gnb = grid_gaussianNB.best_estimator_ # Getting the best estimators
gnb_pred = grid_gaussianNB.predict(X_test) # Predicting y_test by using the best model

# Results
print("Best parameters: ", grid_gaussianNB.best_params_)
print("Best CV score: ", round(grid_gaussianNB.best_score_,2))
print(f"Accuracy :{round(accuracy_score(y_test, gnb_pred),2)}")

# Classification Report and Confusion Matrix
print("\nClassification Report:\n", classification_report(y_test, gnb_pred, target_names = ["Good (0)", "Bad (1)"]))

cm = confusion_matrix(y_test, gnb_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix = cm,
    display_labels = ["Good (1)", "Bad (1)"]
)

disp.plot(cmap = "Blues", values_format = "d")
plt.grid(False)
plt.show()
```

```python
# 1. Importing the libraries
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 2. Importing the model
rfc_model = RandomForestClassifier(random_state=42)

# 3. Setting grid parameter
rfc_param = {
    'n_estimators': [100, 200, 300],
    'max_features': ['sqrt', 'log2'],
    'bootstrap': [True, False]
}

# Building the GridSearch CV 
grid_rfc = GridSearchCV(estimator= rfc_model,
                               param_grid= rfc_param,
                               cv=5,
                               scoring='recall')

# Training the model
grid_rfc.fit(X_train,y_train)

# Obtainig the best estimator and predicing y_test
best_ada = grid_rfc.best_estimator_ # Getting the best estimators
rfc_pred = grid_rfc.predict(X_test) # Predicting y_test by using the best model

# Results
print("Best parameters: ", grid_rfc.best_params_)
print("Best CV score: ", round(grid_rfc.best_score_,2))
print(f"Accuracy :{round(accuracy_score(y_test, rfc_pred),2)}")

# Optional: more evaluation
print("\nClassification Report:\n", classification_report(y_test, rfc_pred, target_names = ["Good (0)", "Bad (1)"]))

cm = confusion_matrix(y_test, rfc_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix = cm,
    display_labels = ["Good (0)", "Bad (1)"]
)

disp.plot(cmap = "Blues", values_format = "d")
plt.grid(False)
plt.show()
```

```python
import joblib
import numpy as np
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI(
    title="API - Testing Credit Risk",
    description="The purpose of this API is to predict if a customer is reliable for taking loans",
)

# Cargar el modelo entrenado con 36 variables
model = joblib.load("credit_model.pkl")


class CustomerInput(BaseModel):
    features: list[float]


@app.get("/")
def home():
    return {"Message": "Credit Risk API is active. Go to /docs for testing."}


@app.post("/predict")
def prediction(customer: CustomerInput):

    # Validar que lleguen exactamente las 36 columnas dummy/procesadas
    if len(customer.features) != 36:
        return {
            "error": f"36 characteristics were expected, but received {len(customer.features)}."
        }

    customer_data = np.array(customer.features).reshape(1, -1)

    pred = int(model.predict(customer_data)[0])
    probabilities = model.predict_proba(customer_data)[0]

    good_probability = float(probabilities[0])
    bad_probability = float(probabilities[1])

    results = "Bad (High risk)" if pred == 1 else "Good (Low risk)"
    approve_loan = bool(pred == 0)

    return {
        "Prediction": pred,
        "Result": results,
        "Good_probability": round(good_probability, 4),
        "Bad_probability": round(bad_probability, 4),
        "Approve_loan": approve_loan,
    }
```
